```python
#numpy.diag() funstion
import numpy as np
```

```python
num=np.matrix([[1,2,3],[4,5,6],[7,8,9]])
```

```python
num
```

```python
type(num)
```

```python
print("main diagonal of the matrix is:",np.diag(num))
print("diagonal above the main diagonal is:",np.diag(num,1))
print("diagonal below the main diagonal is:",np.diag(num,-1))
print("diagonal above the main diagonal is:",np.diag(num,2,2))
print("diagonal below the main diagonal is:",np.diag(num,-2,-2))
```
