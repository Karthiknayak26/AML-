```python

# 1. Create a Pandas Series containing the marks: 78, 85, 92, 67, 88

#     Display:

# 	Complete Series
# 	Values
# 	Index
# 	Data type
# 	Mean
# 	Maximum
# 	Minimum
import pandas as pd 
data=pd.Series([78, 85, 92, 67, 88])
print("complete_series",data)
print("values",data.values)
print("index",data.index)
print("data type",data.dtype)
print("mean",data.mean)
print("maximum",data.max)
print("minimum",data.min)
```

```python
# 2. Create a Series containing:
 
#    	| Student | Marks |
# 	| ------- | ----: |
# 	| Asha    |    85 |
# 	| Ravi    |    78 |
# 	| Meena   |    92 |
# 	| Kiran   |    74 |
# 	| Arun    |    88 |
# 	Display the marks of Meena.



marks = pd.Series(
    [85, 78, 92, 74, 88],
    index=['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun']
)

# Display the marks of Meena
print(marks['Meena'])
```

```python
# 3. Create the following DataFrame:

# | Name  | Age | Department | Marks |
# | ----- | --: | ---------- | ----: |
# | Asha  |  22 | CSE        |    85 |
# | Ravi  |  23 | ECE        |    78 |
# | Meena |  21 | CSE        |    92 |
# | Kiran |  24 | ME         |    74 |
# | Arun  |  22 | ECE        |    88 |

# Display:

# 	Complete DataFrame
# 	First 3 rows
# 	Last 2 rows
# 	Shape
# 	Column names
# 	Data types
df = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, 21, 24, 22],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE"],
    "Marks": [85, 78, 92, 74, 88]
})

print("Complete DataFrame")
print(df)
print("\nFirst 3 rows")
print(df.head(3))
print("\nLast 2 rows")
print(df.tail(2))
print("\nShape")
print(df.shape)
print("\nColumn names")
print(df.columns.tolist())
print("\nData types")
print(df.dtypes)
```

```python
# 4. Using the previous DataFrame, display:

# 	info()
# 	describe()
# 	Number of rows
# 	Number of columns
print("DataFrame info:")
df.info()

print("\nDataFrame describe:")
print(df.describe())

print("\nNumber of rows:", len(df))
print("Number of columns:", len(df.columns))
```

```python
# 5. Using df, display:
# Name column
# Marks column
# Name and Marks together

print("Name column")
print(df['Name'])

print("\nMarks column")
print(df['Marks'])

print("\nName and Marks together")
print(df[['Name', 'Marks']])
```

```python
# 6. Display:
#     First row using iloc
#     Third row using iloc
#     Rows 1 to 3
#     Name and Marks columns using loc

print("First row using iloc:")
print(df.iloc[0])

print("\nThird row using iloc:")
print(df.iloc[2])

print("\nRows 1 to 3:")
print(df.iloc[1:4])

print("\nName and Marks columns using loc:")
print(df.loc[:, ["Name", "Marks"]])
```

```python

# 7. Find:
# 	Students scoring above 80.
# 	Students scoring between 75 and 90.
# 	Students belonging to CSE.
# 	CSE students scoring above 80.

# Students scoring above 80
above_80 = df[df["Marks"] > 80]
print("Students scoring above 80:")
print(above_80[["Name", "Marks"]])

# Students scoring between 75 and 90
between_75_90 = df[(df["Marks"] >= 75) & (df["Marks"] <= 90)]
print("\nStudents scoring between 75 and 90:")
print(between_75_90[["Name", "Marks"]])

# Students belonging to CSE
cse_students = df[df["Department"] == "CSE"]
print("\nStudents belonging to CSE:")
print(cse_students[["Name", "Department"]])

# CSE students scoring above 80
cse_high = df[(df["Department"] == "CSE") & (df["Marks"] > 80)]
print("\nCSE students scoring above 80:")
print(cse_high[["Name", "Department", "Marks"]])
```

```python
# 8. Display students belonging to either CSE or ECE.
Cse_Ece=df[(df['Department']=="CSE") | (df['Department']=="ECE")]
print(Cse_Ece[["Name"]])
```

```python

# <!-- # 9. Add a new column called Result. Students scoring 75 or above should receive #"Pass"otherwise "Fail". -->
df['Result'] = df['Marks'].apply(lambda x: 'Pass' if x >= 75 else 'Fail')
print(df)
```

```python
# 10. Assign grades according to the given scale
df['Grade'] = df['Marks'].apply(lambda x: 'A' if x >= 90 else ('B' if x >= 80 else ('C' if x >= 70 else 'D')))
print(df)
```

```python
# 11. Sort students:
# By marks in ascending order
print("By marks in ascending order:")
print(df.sort_values('Marks'))

# By marks in descending order
print("\nBy marks in descending order:")
print(df.sort_values('Marks', ascending=False))

# By Department and then Marks
print("\nBy Department and then Marks:")
print(df.sort_values(['Department', 'Marks']))
```

```python
# <!-- 12. Consider:

# data = {
#     "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
#     "Age": [22, 23, np.nan, 24, 22],
#     "Marks": [85, np.nan, 92, 74, 88]
# }

# students = pd.DataFrame(data)

# Find:

# Missing values
# Number of missing values in each column
# Rows containing missing values -->
import numpy as np

# Create the students DataFrame with missing values
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

# Missing values
print("Missing values:")
print(students.isnull())

# Number of missing values in each column
print("\nNumber of missing values in each column:")
print(students.isnull().sum())

# Rows containing missing values
print("\nRows containing missing values:")
print(students[students.isnull().any(axis=1)])
```

```python
# 13. Replace:

# 	Missing Age with average Age
# 	Missing Marks with average Marks
# Replace missing Age with average Age
students['Age'].fillna(students['Age'].mean(), inplace=True)

# Replace missing Marks with average Marks
students['Marks'].fillna(students['Marks'].mean(), inplace=True)

print(students)
```

```python
# 14. Remove rows containing missing values.
students_clean = students.dropna()
print(students_clean)
```
