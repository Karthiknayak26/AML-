```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, accuracy_score

# 1. LOAD DATA & SELECT FEATURES
df = pd.read_csv('Social_Network_Ads - Social_Network_Ads.csv')

# Look closely here: We select column index 2 (Age) and 3 (EstimatedSalary)
X = df.iloc[:, [2, 3]].values  
# y is our target (Purchased)
y = df.iloc[:, -1].values      

# 2. SPLIT INTO TRAIN & TEST (75% Train, 25% Test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

# 3. FEATURE SCALING (Make Age and Salary comparable)
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

```

```python
X_train
```

```python
X_test
```

```python
# 1. fitting decision tree classification to the training set
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion='entropy', random_state=0)

# 2. FIXED LINE: Capital X_train, and y_train
classifier.fit(X_train, y_train) 
```

```python
# Force the model to guess the answers for the test data
y_pred = classifier.predict(X_test)
```

```python
#making the confusion matrix
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
Accuracy = (cm[0][0] + cm[1][1]) / cm.sum()
```

```python
from sklearn import metrics
prediction=metrics.accuracy_score(y_test,y_pred)
print("accuracy",prediction*100,'%')
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
import numpy as np
new_individual=np.array([[60,50000]])
new_individual_scaled=sc.transform(new_individual)
prediction=classifier.predict(new_individual_scaled)
if prediction[0]==1:
    print("THE INDIVIDUAL IS PRIDICTED TO PURPHASE")
else:
    print("THE INDIVIDUAL IS PREDICTED NOT TO PURCHASE")        
```
