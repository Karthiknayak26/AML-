```python
#8. Given:
#salary = np.array([25000, 32000, 28000, 45000,50000, 38000, 42000, 30000])
#Find:
#Total salary
#Average salary
#Highest salary
#Lowest salary
#Median salary
#Standard deviation

import numpy as np

salary = np.array([25000, 32000, 28000, 45000, 50000, 38000, 42000, 30000])

print(salary.sum())
print(salary.mean())
print(salary.max())
print(salary.min())
print(np.median(salary))
print(salary.std())
```

```python

#9. Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
#Using NumPy boolean indexing, find:
#Marks greater than 75
#Marks between 50 and 80
#Number of students who scored below 50
#Percentage of students who scored 60 or above
#Do not use a Python for loop.


import numpy as np

marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

greater_than_75 = marks > 75
between_50_and_80 = (marks >= 50) & (marks <= 80)
below_50 = marks < 50
score_60_or_above = marks >= 60

print(marks[greater_than_75])
print(marks[between_50_and_80])
print(np.count_nonzero(below_50))
print((np.count_nonzero(score_60_or_above) / len(marks)) * 100)
```

```python

#10. The following array contains marks of four students in three subjects:
#marks = np.array([
#    [80, 75, 90],
#    [65, 70, 72],
##   [55, 60, 58]
#])

#Calculate:

#Total marks of each student
#Average marks of each student
#Average marks in each subject
#Student with the highest total
#Highest mark in each subject.


marks = np.array([
   [80, 75, 90],
   [65, 70, 72],
   [55, 60, 58]
])

print("total marks of each student",np.sum(marks,axis=1))
print("average marks of each student",np.mean(marks,axis=1))
print("average marks of each student",np.mean(marks,axis=0))
print("student with highest total ",np.max(marks,axis=1))
print("student with highest marks",np.max(marks,axis=0))
```
