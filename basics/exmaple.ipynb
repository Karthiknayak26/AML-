```python
sales_record={
    'price':120,
    'items':4,
    'person':'karthik',
    'email':'karthik@mail.com'
}
```

```python
sales_statements='{} will buy {} items at a price of {}   '
print(sales_statements.format(sales_record['person'],sales_record['items'],sales_record['price']))
```

```python
sales_statements='{} will buy {} items at a price of {} and total price is {} '
print(sales_statements.format(sales_record['person'],sales_record['items'],sales_record['price'],sales_record['items']*sales_record['price']))
```
