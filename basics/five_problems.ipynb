```python
#complete the square numbers using mapping and lambda function
num=[1,2,3,4,5]
squared_numbers=list(map(lambda x:x**2,num))
print(squared_numbers)
```

```python
#compute square of even numbers using list comprehension technique
num=[1,2,3,4,5]
squared_even_numbers=[x**2 for x in num if x%2==0]
```

```python
#implementation of the quicksort algorithm using python 
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)
```
