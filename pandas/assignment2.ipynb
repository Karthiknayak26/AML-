```python

<!-- 17. Consider:
Find:
Average marks by department
Maximum marks by department
Number of students in each department

For each department, calculate:
Mean
Minimum
Maximum
Count -->

import pandas as pd

# Create DataFrame
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun", "Divya"],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE"],
    "Marks": [85, 78, 92, 74, 88, 80]
}

df = pd.DataFrame(data)


avg_marks = df.groupby("Department")["Marks"].mean()
print("--- Average Marks by Department ---")
print(avg_marks)

max_marks = df.groupby("Department")["Marks"].max()
print("\n--- Maximum Marks by Department ---")
print(max_marks)


student_count = df.groupby("Department")["Name"].count()
print("\n--- Number of Students in Each Department ---")
print(student_count)


dept_summary = df.groupby("Department")["Marks"].agg(
    Mean='mean',
    Minimum='min',
    Maximum='max',
    Count='count'
)

print("\n--- Department Summary (Mean, Min, Max, Count) ---")
print(dept_summary)
```

```python
# 19. Use the following dataset:

# data = {
#     "Name": [
#         "Asha", "Ravi", "Meena",
#         "Kiran", "Arun", "Divya",
#         "John", "Priya"
#     ],

#     "Department": [
#         "CSE", "ECE", "CSE",
#         "ME", "ECE", "CSE",
#         "ME", "ECE"
#     ],

#     "Attendance": [
#         92, 75, 96, 68,
#         82, 88, 72, 95
#     ],

#     "Internal": [
#         85, 72, 94, 65,
#         80, 86, 70, 91
#     ],

#     "FinalMarks": [
#         88, 74, 96, 62,
#         83, 89, 68, 93
#     ]
# }

# df = pd.DataFrame(data)


# Perform the following analysis:

# Display the first five records.
# Find the dimensions of the dataset.
# Find the average final mark.
# Find the highest-scoring student.
# Find the lowest-scoring student.
# Display students scoring above 80.
# Find average marks by department.
# Find students having attendance above 80%.
# Sort students by FinalMarks.
# Add a Result column: Pass if FinalMarks ≥ 70, otherwise Fail.
# Add a performance category:
# 	≥90 → Excellent
# 	≥80 → Very Good
# 	≥70 → Good
# 	<70 → Needs Improvement
# Find the department having the highest average FinalMarks.


import pandas as pd

# Create DataFrame
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun", "Divya", "John", "Priya"],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE", "CSE", "ME", "ECE"],
    "Attendance": [92, 75, 96, 68, 82, 88, 72, 95],
    "Internal": [85, 72, 94, 65, 80, 86, 70, 91],
    "FinalMarks": [88, 74, 96, 62, 83, 89, 68, 93]
}

df = pd.DataFrame(data)

# 1. Display the first five records
print("1. First 5 records:")
print(df.head())

# 2. Find the dimensions of the dataset (rows, columns)
print("\n2. Dimensions of dataset (Rows, Columns):")
print(df.shape)

# 3. Find the average final mark
avg_final = df["FinalMarks"].mean()
print(f"\n3. Average Final Mark: {avg_final:.2f}")

# 4. Find the highest-scoring student
highest_student = df.loc[df["FinalMarks"].idxmax()]
print(f"\n4. Highest-scoring student: {highest_student['Name']} ({highest_student['FinalMarks']} marks)")

# 5. Find the lowest-scoring student
lowest_student = df.loc[df["FinalMarks"].idxmin()]
print(f"\n5. Lowest-scoring student: {lowest_student['Name']} ({lowest_student['FinalMarks']} marks)")

# 6. Display students scoring above 80
print("\n6. Students scoring FinalMarks > 80:")
print(df[df["FinalMarks"] > 80][["Name", "FinalMarks"]])

# 7. Find average marks by department
print("\n7. Average FinalMarks by Department:")
dept_avg = df.groupby("Department")["FinalMarks"].mean()
print(dept_avg)

# 8. Find students having attendance above 80%
print("\n8. Students with Attendance > 80%:")
print(df[df["Attendance"] > 80][["Name", "Attendance"]])

# 9. Sort students by FinalMarks (descending order)
print("\n9. Students sorted by FinalMarks (Highest to Lowest):")
print(df.sort_values(by="FinalMarks", ascending=False))

# 10. Add a Result column: Pass if FinalMarks >= 70, otherwise Fail
df["Result"] = df["FinalMarks"].apply(lambda x: "Pass" if x >= 70 else "Fail")

# 11. Add a performance category
def get_performance(marks):
    if marks >= 90:
        return "Excellent"
    elif marks >= 80:
        return "Very Good"
    elif marks >= 70:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance"] = df["FinalMarks"].apply(get_performance)

print("\n10 & 11. DataFrame with 'Result' and 'Performance' columns:")
print(df)

# 12. Find the department having the highest average FinalMarks
best_dept = dept_avg.idxmax()
best_dept_score = dept_avg.max()
print(f"\n12. Department with highest average FinalMarks: {best_dept} ({best_dept_score:.2f})")
```

```python
# 20. Create:

# employees = pd.DataFrame({
#     "Employee": [
#         "E01", "E02", "E03", "E04",
#         "E05", "E06", "E07", "E08"
#     ],

#     "Department": [
#         "IT", "HR", "IT", "Finance",
#         "HR", "Finance", "IT", "Finance"
#     ],

#     "Experience": [
#         5, 3, 7, 4,
#         6, 8, 2, 5
#     ],

#     "Salary": [
#         60000, 45000, 75000, 65000,
#         55000, 85000, 48000, 70000
#     ]
# })


# Determine:

# Average salary.
# Maximum and minimum salary.
# Employee receiving the highest salary.
# Employees earning above ₹60,000.
# Average salary for each department.
# Average experience by department.
# Number of employees in each department.
# Sort employees by salary.
# Find IT employees with more than 3 years of experience.
# Add a SalaryLevel column:
# 	≥ ₹75,000 → High
# 	₹50,000–₹74,999 → Medium
# 	< ₹50,000 → Low

import pandas as pd

# Create DataFrame
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],
    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],
    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

# 1. Average salary
avg_salary = employees["Salary"].mean()
print(f"1. Average Salary: ₹{avg_salary:,.2f}")

# 2. Maximum and minimum salary
max_salary = employees["Salary"].max()
min_salary = employees["Salary"].min()
print(f"2. Maximum Salary: ₹{max_salary:,}")
print(f"   Minimum Salary: ₹{min_salary:,}")

# 3. Employee receiving the highest salary
highest_earner = employees.loc[employees["Salary"].idxmax()]
print(f"3. Highest Earner: Employee {highest_earner['Employee']} with ₹{highest_earner['Salary']:,}")

# 4. Employees earning above ₹60,000
print("\n4. Employees earning above ₹60,000:")
print(employees[employees["Salary"] > 60000][["Employee", "Department", "Salary"]])

# 5. Average salary for each department
print("\n5. Average Salary by Department:")
print(employees.groupby("Department")["Salary"].mean())

# 6. Average experience by department
print("\n6. Average Experience by Department (in years):")
print(employees.groupby("Department")["Experience"].mean())

# 7. Number of employees in each department
print("\n7. Number of employees in each department:")
print(employees["Department"].value_counts())

# 8. Sort employees by salary (Descending order)
print("\n8. Employees sorted by Salary (Highest to Lowest):")
print(employees.sort_values(by="Salary", ascending=False))

# 9. Find IT employees with more than 3 years of experience
print("\n9. IT employees with > 3 years of experience:")
it_exp = employees[(employees["Department"] == "IT") & (employees["Experience"] > 3)]
print(it_exp)

# 10. Add SalaryLevel column
def classify_salary(salary):
    if salary >= 75000:
        return "High"
    elif salary >= 50000:
        return "Medium"
    else:
        return "Low"

employees["SalaryLevel"] = employees["Salary"].apply(classify_salary)

print("\n10. DataFrame with 'SalaryLevel' column:")
print(employees)

```
