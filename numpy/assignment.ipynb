```python
# 3. Create a NumPy array containing the numbers from 10 to 50.
# Display: the array, number of elements, shape, dimension, and data type

import numpy as np
arr = np.arange(10, 51)

print("The array:", arr)
print("Number of elements:", arr.size)
print("Shape:", arr.shape)
print("Dimension:", arr.ndim)
print("Data type:", arr.dtype)
```
