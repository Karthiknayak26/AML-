```python

#9. Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
#Using NumPy boolean indexing, find:
#Marks greater than 75
#Marks between 50 and 80
#Number of students who scored below 50
#Percentage of students who scored 60 or above
#Do not use a Python for loop.


import numpy as np

marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

greater_than_75 = marks > 75
between_50_and_80 = (marks >= 50) & (marks <= 80)
below_50 = marks < 50
score_60_or_above = marks >= 60

print(marks[greater_than_75])
print(marks[between_50_and_80])
print(np.count_nonzero(below_50))
print((np.count_nonzero(score_60_or_above) / len(marks)) * 100)
```

```python

# 11. Create:

# A = np.array([
#     [1, 2],
#     [3, 4]
# ])

# B = np.array([
#     [5, 6],
#     [7, 8]
# ])

# Perform:

# Matrix addition
# Element-wise multiplication
# Matrix multiplication
# Transpose of A.

# explain the difference between:
# A * B and A @ B

import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A + B =")
print(A + B)

print("\nA * B (element-wise multiplication) =")
print(A * B)

print("\nA @ B (matrix multiplication) =")
print(A @ B)

print("\nTranspose of A =")
print(A.T)

print("\nDifference between A * B and A @ B:")
print("A * B multiplies matching elements position by position.")
print("A @ B performs matrix multiplication using dot products of rows and columns.")
print("So, A * B keeps the same shape as A and B, while A @ B follows matrix multiplication rules and requires compatible dimensions.")
```

```python
# 12. Daily temperatures for two weeks are:

# temperature = np.array([
#     31, 32, 33, 35, 36, 34, 32,
#     30, 29, 31, 33, 35, 37, 36
# ])

# Find:

# Mean temperature
# Maximum and minimum temperatures
# Number of days above the average temperature
# Temperatures greater than 34°C
# Difference between maximum and minimum
# Standard deviation

import numpy as np

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

mean_temp = temperature.mean()
max_temp = temperature.max()
min_temp = temperature.min()
above_avg = temperature[temperature > mean_temp]
above_34 = temperature[temperature > 34]
diff = max_temp - min_temp
std_dev = temperature.std()

print("Mean temperature:", mean_temp)
print("Maximum temperature:", max_temp)
print("Minimum temperature:", min_temp)
print("Number of days above average:", len(above_avg))
print("Temperatures greater than 34°C:", above_34)
print("Difference between maximum and minimum:", diff)
print("Standard deviation:", std_dev)
```

```python
# 13. Monthly sales of a company are:
# sales = np.array([
#     12000, 15000, 17000, 14000,
#     18000, 21000, 19000, 22000,
#     25000, 23000, 26000, 28000
# ])

# Determine:

# Total annual sales
# Average monthly sales
# Month with the highest sales
# Month with the lowest sales
# Number of months with sales above ₹20,000
# Percentage increase from January to December

# Hint: Remember that NumPy indexing starts at 0, while month numbering normally starts at 1.

import numpy as np

sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

# Total annual sales
annual_total = sales.sum()

# Average monthly sales
average_sales = sales.mean()

# Month with the highest sales
highest_index = sales.argmax()
highest_month = highest_index + 1  # convert 0-based index to 1-based month number

# Month with the lowest sales
lowest_index = sales.argmin()
lowest_month = lowest_index + 1

# Number of months with sales above ₹20,000
above_20000 = sales[sales > 20000]
count_above_20000 = len(above_20000)

# Percentage increase from January to December
january = sales[0]
december = sales[-1]
percentage_increase = ((december - january) / january) * 100

print("Total annual sales:", annual_total)
print("Average monthly sales:", average_sales)
print("Month with the highest sales:", highest_month)
print("Month with the lowest sales:", lowest_month)
print("Number of months with sales above ₹20,000:", count_above_20000)
print("Percentage increase from January to December:", percentage_increase, "%")
```

```python

# 14. Consider marks of five students in four subjects:

# marks = np.array([
#     [78, 85, 90, 88],
#     [65, 70, 72, 68],
#     [92, 88, 95, 91],
#     [55, 60, 58, 62],
#     [80, 75, 82, 79]
# ])

# Write a NumPy program that calculates:

# Total marks of every student
# Average of every student
# Average mark for every subject
# Highest-scoring student
# Lowest-scoring student
# Students whose average is greater than 80
# Overall class average

# Then assign a performance category based on average:

# |  Average | Category          |
# | -------: | ----------------- |
# |     ≥ 85 | Excellent         |
# | 70–84.99 | Good              |
# | 50–69.99 | Average           |
# |     < 50 | Needs Improvement |

# Challenge: Try assigning these categories using np.where() or np.select() instead of a Python loop.

import numpy as np

marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

student_totals = marks.sum(axis=1)
student_averages = marks.mean(axis=1)
subject_averages = marks.mean(axis=0)

highest_student = np.argmax(student_averages) + 1
lowest_student = np.argmin(student_averages) + 1
students_above_80 = np.where(student_averages > 80)[0] + 1
overall_class_average = marks.mean()

print("Total marks of every student:", student_totals)
print("Average of every student:", student_averages)
print("Average mark for every subject:", subject_averages)
print("Highest-scoring student:", highest_student)
print("Lowest-scoring student:", lowest_student)
print("Students whose average is greater than 80:", students_above_80)
print("Overall class average:", overall_class_average)

# Performance category using np.select()
conditions = [
    student_averages >= 85,
    (student_averages >= 70) & (student_averages < 85),
    (student_averages >= 50) & (student_averages < 70),
    student_averages < 50
]
labels = ["Excellent", "Good", "Average", "Needs Improvement"]
performance_categories = np.select(conditions, labels, default="Needs Improvement")

print("Performance categories:", performance_categories)
```

```python
# 21. Take your own example to normalize an array.  
import numpy as np

# Own example array
arr = np.array([10, 20, 30, 40, 50])

# Minimum and maximum values
min_val = arr.min()
max_val = arr.max()

# Min-Max normalization formula
normalized = (arr - min_val) / (max_val - min_val)

print("Original array:", arr)
print("Normalized array:", normalized)
```
